```python
# Experiment 9
# Roll No: 24BAD017
# Transformer-Based LLM Chatbot

import torch
from transformers import AutoTokenizer, AutoModelForCausalLM

model_name = "distilgpt2"

tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForCausalLM.from_pretrained(model_name)

tokenizer.pad_token = tokenizer.eos_token

print("Roll No: 24BAD017")
print("Pre-trained LLM loaded successfully!")
print("Model Name:", model_name)
```

```python
prompt = "Artificial Intelligence is"

inputs = tokenizer(prompt, return_tensors="pt")

outputs = model.generate(
    inputs["input_ids"],
    attention_mask=inputs["attention_mask"],
    max_new_tokens=50,
    temperature=0.7,
    do_sample=True,
    pad_token_id=tokenizer.eos_token_id
)

response = tokenizer.decode(outputs[0], skip_special_tokens=True)

print("User Prompt:", prompt)
print("Generated Response:", response)
print("Darsha Padmanaban 24BAD017")
```

```python
chat_history = []

def chatbot_response(user_input):
    global chat_history

    conversation = ""

    for user, bot in chat_history[-3:]:
        conversation += "User: " + user + "\n"
        conversation += "Bot: " + bot + "\n"

    prompt = conversation + "User: " + user_input + "\nBot:"

    inputs = tokenizer(
        prompt,
        return_tensors="pt",
        truncation=True,
        max_length=700
    )

    outputs = model.generate(
        inputs["input_ids"],
        attention_mask=inputs["attention_mask"],
        max_new_tokens=80,
        temperature=0.7,
        do_sample=True,
        pad_token_id=tokenizer.eos_token_id
    )

    new_tokens = outputs[0][inputs["input_ids"].shape[1]:]

    response = tokenizer.decode(
        new_tokens,
        skip_special_tokens=True
    ).strip()

    chat_history.append((user_input, response))

    return response

print("Chatbot function created successfully!")
print("Roll No: 24BAD017")
```

```python
print("User: What is Artificial Intelligence?")
print("Chatbot:", chatbot_response("What is Artificial Intelligence?"))

print("\nUser: What are its applications?")
print("Chatbot:", chatbot_response("What are its applications?"))
```

```python
import gradio as gr

def respond(message, history):
    prompt = "Question: " + message + "\nAnswer:"

    inputs = tokenizer(prompt, return_tensors="pt")

    outputs = model.generate(
        inputs["input_ids"],
        max_new_tokens=80,
        do_sample=True,
        temperature=0.7,
        top_p=0.9,
        repetition_penalty=1.3,
        no_repeat_ngram_size=3,
        pad_token_id=tokenizer.eos_token_id
    )

    answer = tokenizer.decode(
        outputs[0][inputs["input_ids"].shape[1]:],
        skip_special_tokens=True
    ).strip()

    return answer

demo = gr.ChatInterface(
    fn=respond,
    title="LLM Chatbot - 24BAD017",
    description="Transformer-Based Chatbot using DistilGPT2"
)

demo.launch(share=True)
```
